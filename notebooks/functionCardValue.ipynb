### Card Value Function  
Run the function and enter a card to see its price from multiple providers (if available) and the average market price for that card per UUID.

```python
import pandas as pd

# Ensure no outout wrapping.
pd.set_option("display.expand_frame_repr", False)

# Display all in output.
pd.set_option('display.max_rows', None)

dfm3 = pd.read_csv('../data/dataMagic/cleanMagicIndPrices.csv')
# dfm3 = pd.read_pickle("dfm3.pkl")
```

```python
def cardValue():
    # Requires user input and strips unnecessary spaces.
    cardName = input("Please enter MTG card name (or type esc to cancel): ").strip()
    if cardName.lower() == "esc":
        return None
    
    # Requires valid card name.
    while not (dfm3["name"].str.lower() == cardName.lower()).any():
        cardName = input("Please enter a valid card name (or type esc to exit): ").strip()
        if cardName.lower() == "esc":
            return None
    
    rows = dfm3[dfm3["name"].str.lower() == cardName.lower()]
    subset = rows[["name", "setName", "priceProvider", "price",
                   "avgMarketPrice", "cardFinish", "uuid"]]

    return subset
```

```python
# Run the function.
# If example card names are needed: Venom, Aladdin, Buster Sword
cardValue()
```
