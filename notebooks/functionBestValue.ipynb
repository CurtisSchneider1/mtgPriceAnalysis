### Best Value Function  
This is great for determining where to best spend money purely based on card value.  
Disclaimer: There is a chance that providers or related sellers may post a card's price or information erroneously and should be verified individually.

```python
import pandas as pd

# Display all in output.
pd.set_option('display.max_rows', None)

dfm3 = pd.read_csv('../data/dataMagic/cleanMagicIndPrices.csv')
# dfm3 = pd.read_pickle("../data/dataMagic/dfm3.pkl")
```

```python
def bestValue(threshold = 0.96):
    """
    This function outputs cards from dfm3 that meet the threshold.

    Args:
        float: Threshold percentage

    Returns:
        values: Those cards whose prices meet the threshold percentage between price and average market price.
    """
    result = dfm3['price'] <= dfm3['avgMarketPrice'] * (1 - threshold)
    
    subset = dfm3.loc[result, ["name", "setName", "priceProvider", "price",
            "avgMarketPrice", "cardFinish", "uuid"]]
    
    return subset
```

```python
# Default is 96%.
bestValue(0.97)
```
