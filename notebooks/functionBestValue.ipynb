### Function to Show Best Value

```python
import pandas as pd

# Display all in output.
pd.set_option('display.max_rows', None)

dfm3 = pd.read_pickle("dfm3.pkl")
```

```python
def bestValue(threshold = 0.90):
    """
    Function to return cards that are undervalued compared to market price.
    Can accept different threshold percentages when calling the function.
    """
    result = dfm3['price'] <= dfm3['avgMarketPrice'] * (1 - threshold)
    
    subset = dfm3.loc[result, ["name", "setName", "priceProvider", "price",
            "avgMarketPrice", "cardFinish", "uuid"]]
    
    return subset
```

```python
# Default is 90%.
bestValue(0.97)
```
