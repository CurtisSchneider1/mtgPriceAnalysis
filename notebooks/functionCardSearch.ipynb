### Card Value Function  
Run the function and enter a card name to see its price from multiple providers (if available) and the average market price for that card per UUID.

```python
import pandas as pd

# Ensure no output wrapping.
pd.set_option("display.expand_frame_repr", False)

# Display all rows in output.
pd.set_option('display.max_rows', None)

dfm3 = pd.read_csv('../data/dataMagic/cleanMagicIndPrices.csv')
# dfm3 = pd.read_pickle("../data/dataMagic/dfm3.pkl")
```

```python
def cardValue():
    """ 
    Search dfm3 for all instances of an individual card.

    Args:
        str: User inputted card name.

    Returns:
        values: Grouped card and price information when card name matches input.
    """
  
    # Requires user input and strips unnecessary spaces.
    cardName = input("Please enter MTG card name (or type esc to cancel search): ").strip()
    if cardName.lower() == "esc":
        return None
    
    # Requires valid card name.
    while not (dfm3["name"].str.lower() == cardName.lower()).any():
        cardName = input("Please enter a valid card name (or type esc to exit): ").strip()
        if cardName.lower() == "esc":
            return None
    
    rows = dfm3[dfm3["name"].str.lower() == cardName.lower()]
    subset = rows[["name", "setName", "priceProvider", "price",
                   "avgMarketPrice", "cardFinish", "uuid"]]

    # Group results by cardFinish per UUID.
    subset = subset.sort_values(
        by=["uuid", "cardFinish", "priceProvider"],
        # Key transforms the column before sorting.
        key=lambda col: (
            # Converts text to numbers for custom mapping because 'foil' comes before 'normal' alphabetically.
            # If column name = cardFinish, then apply logic.
            # .fillna fills any potential NaN results by putting the original string value back, just in case.
            col.map({"normal": 0, "foil": 1}).fillna(col) if col.name == "cardFinish" else col
        )
    )
    
    return subset
```

```python
# Run the function.
# If example card names are needed: Venom, Aladdin, Buster Sword
cardValue()
```
